## Principal Component Analysis - City Quality of Life

### Aim

To apply Principal Component Analysis (PCA) to a City Quality of Life dataset and identify the variables with the highest contribution to each principal component.

```python
# ------------------------------------------------------------
# 1. Import required libraries
# ------------------------------------------------------------

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# ------------------------------------------------------------
# 2. Create the PCA Loading Dataset
# ------------------------------------------------------------

data = {
    "Variable": [
        "Climate",
        "Housing",
        "Health",
        "Crime",
        "Transportation",
        "Education",
        "Arts",
        "Recreation",
        "Economy"
    ],

    "PC1": [
        0.190,
        0.544,
        0.782,
        0.365,
        0.585,
        0.394,
        0.985,
        0.520,
        0.142
    ],

    "PC2": [
        0.017,
        0.020,
        -0.605,
        0.294,
        0.085,
        -0.273,
        0.126,
        0.402,
        0.150
    ],

    "PC3": [
        0.207,
        0.204,
        0.144,
        0.585,
        0.234,
        0.027,
        -0.111,
        0.519,
        0.239
    ]
}

df = pd.DataFrame(data)

print("PCA Loading Table:")
print(df)
```

```python
# ------------------------------------------------------------
# 3. Find the Variable with the Highest Loading for Each PC
# ------------------------------------------------------------

for pc in ["PC1", "PC2", "PC3"]:

    # Find index of largest absolute loading
    index = df[pc].abs().idxmax()

    variable = df.loc[index, "Variable"]
    loading = df.loc[index, pc]

    print("\n", pc)
    print("Variable :", variable)
    print("Loading  :", loading)
```

```python
# ------------------------------------------------------------
# 4. Find the Overall Highest Loading
# ------------------------------------------------------------

pc_columns = ["PC1", "PC2", "PC3"]

max_value = 0
max_variable = ""
max_pc = ""

for pc in pc_columns:

    index = df[pc].abs().idxmax()

    if abs(df.loc[index, pc]) > max_value:

        max_value = abs(df.loc[index, pc])
        max_variable = df.loc[index, "Variable"]
        max_pc = pc

print("\n------------------------------------------------------------")
print("Overall Highest Loading")
print("------------------------------------------------------------")
print("Variable :", max_variable)
print("Principal Component :", max_pc)
print("Loading :", df.loc[
    df["Variable"] == max_variable, max_pc
].values[0])
```

```python
# ------------------------------------------------------------
# 5. Display the Final Answer
# ------------------------------------------------------------

print("\n============================================================")
print("FINAL ANSWER")
print("============================================================")

print("PC1 -> Arts        (Loading = 0.985)")
print("PC2 -> Health      (Loading = -0.605)")
print("PC3 -> Crime       (Loading = 0.585)")

print("\nThe variable with the highest loading overall is:")
print("Arts, with a loading of 0.985 on PC1.")
```
